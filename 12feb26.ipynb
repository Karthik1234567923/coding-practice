This program calculates the perimeter of a rectangle.

It accepts the length and breadth of the rectangle from the user, calculates the perimeter using the formula 2 × (length + breadth), and displays the result.



Input Format

First line: an integer length (length of the rectangle).

Second line: an integer breadth (breadth of the rectangle).



Output Format

Display the perimeter in the format:

Perimeter of the rectangle is: <perimeter>

```python
# Read inputs
print("Enter the length: ")
length = int(input())

print("Enter the breadth: ")
breadth = int(input())

# Calculate perimeter
perimeter = 2 * (length + breadth)

# Display result20
print(f"Perimeter of the rectangle is: {perimeter}")
```
