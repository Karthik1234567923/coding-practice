Prime Number using Function in Python

```python
def check_prime(n):
    if n <= 1:
        return "Not Prime"

    for i in range(2, n):
        if n % i == 0:
            return "Not Prime"

    return "Prime"


num = int(input("Enter a number: "))
result = check_prime(num)
print(num, "is", result)
```
