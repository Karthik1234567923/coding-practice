Problem: Largest of Three Numbers
✅ Logic

Read three numbers as input.

Compare them using if–elif–else conditions.

Print the largest number.

```python
a = int(input())
b = int(input())
c = int(input())

if a >= b and a >= c:
    print(a)
elif b >= a and b >= c:
    print(b)
else:
    print(c)
```
