Temperature Conversion (Both Conditions)
🧠 Logic (Long Explanation)

First, ask the user to choose conversion type:

Enter 1 for Celsius to Fahrenheit

Enter 2 for Fahrenheit to Celsius

If user selects 1:

Ask temperature in Celsius.

Use formula:
F = (C × 9/5) + 32

Print Fahrenheit value.

If user selects 2:

Ask temperature in Fahrenheit.

Use formula:
C = (F − 32) × 5/9

Print Celsius value.

Use if-elif to check user choice.

```python
choice = int(input("Enter 1 for c to f and 2 for f to c"))
if choice ==1:
    c = int(input("Enter temperature in c:"))
    f = (c*9/5)+32
    print(f)
elif choice ==2:
    f = int(input("Enter temperature in f:"))
    c = (f-32)*5/9
    print(c)
else:
    print("Invalied number")
```
