Question 5: Grade Calculation

Problem:
Write a Python program to print the grade based on marks.

Marks	Grade
≥ 90	A
≥ 75	B
≥ 60	C
< 60	Fail

```python
marks = int(input())

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("Fail")
```
