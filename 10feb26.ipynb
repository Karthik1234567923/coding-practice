Write a Python program to print the following number pyramid pattern.

```python
rows = 5

for i in range(1, rows + 1):
    # print spaces
    for j in range(rows - i):
        print(" ", end="")

    # print numbers
    for k in range(1, i + 1):
        print(k, end=" ")

    print()
```
