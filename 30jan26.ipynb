You are developing a Financial Calculation Application for a small business that needs to handle monetary values. In financial applications, it's essential to display prices and amounts accurately, especially when dealing with user input. This application should allow users to input a float value (such as an amount of money), and then it will display the rounded-down and rounded-up values. This functionality is particularly useful in scenarios where pricing strategies might involve rounding amounts to the nearest integer for display purposes, ensuring clarity in transactions. In this scenario, the program prompts the user to input a float value representing a monetary amount. The program then calculates and displays three values: The original float value (truncated to an integer). The rounded-up value (using the ceil() function). The rounded-down value (using the floor() function). This output helps users see both the potential maximum and minimum rounded values associated with a given amount.

```python
import math
num = float(input())
print(int(num))
print(math.ceil(num))
print(math.floor(num))
```

You are tasked with developing a Basic Mathematical Operations Calculator for a finance application. The application allows users to perform some essential operations on two given numbers. These operations include finding the square root, power of a number, and the absolute value of a number. This type of calculator is often used in financial applications where users need to quickly compute mathematical functions like interest calculations, forecasting, or risk analysis. Write a program that takes two numbers as input from the user and performs basic mathematical operations on them, such as: Finding the square root of the first number, Calculating the power of the first number raised to the second number, Determining the absolute value of both numbers. The program should output the results of these mathematical functions.

```python
import math

# Read inputs
num1 = float(input())
num2 = float(input())

# Square root of first number
print("Square Root of first number:", math.sqrt(num1))

# Power calculation
print("First number raised to the power of second number:", math.pow(num1, num2))

# Absolute values
print("Absolute value of first number:", abs(num1))
print("Absolute value of second number:", abs(num2))
```
