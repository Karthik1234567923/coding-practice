Problem Submissions Leaderboard Discussions Imagine you are designing a number pattern for an educational game where the player needs to solve problems based on number patterns. In this game, the player must observe the sequence and predict the numbers in the next row. The pattern described is part of the level progression, where the complexity increases with every row, and the player needs to calculate the next set of digits.

```python
n = int(input())

num = 1

for i in range(1, n + 1):
    for j in range(i):
        print(num, end="")
        num += 1
    print()
```
