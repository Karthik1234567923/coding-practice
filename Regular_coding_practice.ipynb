```python
#What is Python?

Python is a high-level programming language used to tell a computer what to do.
It is easy to read, easy to write, and very powerful.

People use Python for:

🧮 Mathematics & science

🤖 Artificial Intelligence

🌐 Websites

📊 Data analysis

⚙️ Automation

Basic Rules of Python
1. Python is case-sensitive
name = "Karthik"
Name = "Karthik"


These are different variables.

2. Indentation is very important

Python uses spaces to define blocks of code (not { } like other languages).

✅ Correct:

if 5 > 3:
    print("Five is greater")


❌ Wrong:

if 5 > 3:
print("Five is greater")

3. Variables do not need a data type

Python automatically understands the type.

x = 10        # integer
y = 3.14      # float
name = "Ram"  # string

4. Use # for comments

Comments are ignored by Python.

# This is a comment
print("Hello")

5. Statements usually end in a new line
print("Hello")
print("World")

6. Keywords cannot be used as variable names

❌ Wrong:

if = 10


✅ Correct:

number = 10

7. Python uses simple English words

Examples:

if, else

for, while

print, input

8. Quotes matter in strings
print("Hello")
print('Hello')
```

```python
#swap woking rules
*Swapping using a third variable
a = 5
b = 10

temp = a
a = b
b = temp

print("a =", a)
print("b =", b)
```

```python
Swapping without using a third variable
a = 5
b = 10
a,b = b,a
print("a=",a)
print("b=",b)
```

```python
boy_name = input("boy_name>>")
boy_age = int(input("boy_age>>"))
girl_name = input("girl_name>>")
girl_age = int(input("girl_age>>"))
# using abs because sometimes boy might be younger
age_diff = boy_age - girl_age
age_diff = abs(age_diff)
print(boy_name + " loves " + girl_name + "age difference is " + str(age_diff))
print(f"{boy_name} loves {girl_name} age difference is {age_diff}")
```
