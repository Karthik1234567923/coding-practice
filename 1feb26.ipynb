```python

#	Develop  a python program on the given number for performing arithmetic operators


#step 1:input
#we use float()to handle decimal numbers
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Step 2: Processing
sum_val = num1 + num2
diff_val = num1 - num2
prod_val = num1 * num2
div_val = num1 / num2
mod_val = num1 % num2
power_val = num1 ** num2

# Step 3: Output
print("Addition (+):", sum_val)
print("Subtraction (-):", diff_val)
print("Multiplication (*):", prod_val)
print("Division (/):", div_val)
print("Modulus (%):", mod_val)
print("Power (**):", power_val)


outEnter the first number: 7.
Enter the second number: 8
Addition (+): 15.7
Subtraction (-): -0.2999999999999998
Multiplication (*): 61.6
Division (/): 0.9625
Modulus (%): 7.7
Power (**): 12357362.915476812

#velop a python program to find the largest of the given three numbers:



# Step 1: Input
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

# Step 2: Logic using if-elif-else
if a >= b and a >= c:
    print(f"{a}is the largest number.")
elif b >= a and b >= c:
    print(f"{b}is the largest number.")
else:
    print(f"{c}is the largest number.")




	#Develop a python program to convert the given temperatures from celcius to

Code :
farenhit and vice versa
print("1. Celsius to Fahrenheit")
print("2. Fahrenheit to Celsius")

choice = input("Enter choice (1/2): ")

if choice == '1':
    c = float(input("Enter temperature in Celsius: "))
    f = (c * 9/5) + 32
    print(f"{c}°C is equal to {f:.2f}°F")

elif choice == '2':
    f = float(input("Enter temperature in Fahrenheit: "))
    c = (f - 32) * 5/9
    print(f"{f}°F is equal to {c:.2f}°C")

else:
    print("Invalid choice.")





#	Develop a python program to identifying the type of triangle from the given dimensions
# Program to identify the type of triangle

a = float(input("Enter side 1: "))
b = float(input("Enter side 2: "))
c = float(input("Enter side 3: "))

# Check if triangle is valid
if (a + b > c) and (a + c > b) and (b + c > a):
    if a == b == c:
        print("The triangle is Equilateral")
    elif a == b or b == c or a == c:
        print("The triangle is Isosceles")
    else:
        print("The triangle is Scalene")
else:
    print("The given sides do not form a triangle")
output :
Enter side 1: 9
Enter side 2: 7
Enter side 3: 8
The triangle is Scalene

=== Code Execution Successful ==
```
