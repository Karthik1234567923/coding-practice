```python
#pattern printing
n = int(input("Enter the number of rows"))
for i in range(1, n+1):
  for j in range(1, i+1):
    print("*" ,end="")
  print()
```

```python
n = int(input("Enter number of rows: "))
num =1

for i in range(1, n+1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```
