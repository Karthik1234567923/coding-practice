1. Arithmetic Operations
🧠 Logic (Long Explanation):

First, we ask the user to enter two numbers.

Store the numbers in variables (a and b).

Use arithmetic operators:

+ to add the numbers

- to subtract

* to multiply

/ to divide

Print each result separately.

The program simply performs basic mathematical calculations on user input.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
```

2. Largest of Three Numbers
🧠 Logic (Long Explanation):

Ask the user to enter three numbers.

Store them in variables (a, b, c).

Compare the numbers using if-elif-else.

First check if a is greater than both b and c.

If true, a is the largest.

If not, check if b is greater than c.

If true, b is the largest.

Otherwise, c is the largest.

Print the largest number.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a > b and a > c:
    print("Largest number is", a)
elif b > c:
    print("Largest number is", b)
else:
    print("Largest number is", c)
```
