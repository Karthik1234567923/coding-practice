Logic (Long Explanation for Exam)

Ask the user to enter a number (n).

A prime number is a number that:

Is greater than 1

Has only two factors: 1 and itself

Example:
7 → factors are 1 and 7 → Prime
6 → factors are 1, 2, 3, 6 → Not prime

If n is less than or equal to 1 → Not prime.

Otherwise:

Check divisibility from 2 to n-1.

If any number divides n exactly → Not prime.

If no number divides it → Prime.

```python
n = int(input("Enter a number: "))

if n <= 1:
    print("Not a Prime Number")
else:
    for i in range(2, n):
        if n % i == 0:
            print("Not a Prime Number")
            break
    else:
        print("Prime Number")
```
