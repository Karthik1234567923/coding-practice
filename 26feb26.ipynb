Write a program to print a right-angled triangle star pattern based on the given number of rows.

```python
print("Enter number of rows: ")
n = int(input())

print("--- Pattern ---")

for i in range(1, n + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```
